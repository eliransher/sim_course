## `while`

### לולאת while חוזרת על הקוד כל עוד התנאי מתקיים.

### מתאימה כשלא יודעים מראש כמה סיבובים יהיו — למשל, להטיל מטבע עד שמקבלים גם עץ וגם פלי (מחברת 1).

### צריך לוודא שהתנאי יהפוך מתישהו לשקר, אחרת הלולאה לא תיגמר.

#### מופיע במחברות: 1, 4, 6

</div>

```python
np.random.seed(1)
chosen = []
trials = 0
while len(chosen) <= 1:
    trials += 1
    res = np.random.choice(['pali', 'etz'])
    if res not in chosen:
        chosen.append(res)
print('needed', trials, 'tosses:', chosen)
```

## `break  continue`

### הפקודה `break` יוצאת מהלולאה מיד, גם אם הלולאה עוד לא הסתיימה.

### הפקודה `continue` מדלגת על שאר הקוד בסיבוב הנוכחי, וממשיכה לסיבוב הבא.

### למשל: להפסיק סימולציה כשמגיעים לזמן מסוים, או לדלג על לקוחות בתקופת החימום.

</div>

```python
np.random.seed(0)
clock = 0.0
for customer in range(1000):
    clock += np.random.exponential(1.0)   # next arrival time
    if clock > 5:
        break                             # stop: we passed time 5
print('stopped at customer', customer, 'time', round(clock, 2))

for i in range(6):
    if i % 2 == 0:
        continue                          # skip even numbers
    print('odd:', i)
```

## `def  return`

### הגדרת פונקציה: נותנים שם לקטע קוד, כדי שאפשר יהיה להריץ אותו שוב עם קלטים שונים.

### הפקודה `return` מחזירה את התוצאה. אפשר להחזיר כמה ערכים בבת אחת.

### אפשר לתת לפרמטר ערך ברירת מחדל, כמו `mu=1.0` בדוגמה.

#### מופיע במחברות: 1, 4, 5, 6, 10

</div>

```python
def mm1_stats(lam, mu=1.0):
    """Utilization and mean number in system of an M/M/1 queue."""
    rho = lam / mu
    L = rho / (1 - rho)
    return rho, L

rho, L = mm1_stats(0.8)
print(f'rho = {rho}, L = {L:.2f}')
print(mm1_stats(0.5, mu=2.0))
```

## `class`

### מחלקה מאגדת יחד נתונים ופונקציות. במחברת 6 התור כולו בנוי כמחלקה.

### הפונקציה `__init__` רצה כשיוצרים עצם חדש, והמילה `self` מציינת את העצם עצמו.

### אין צורך לכתוב מחלקות בעצמכם בקורס, אבל כדאי להבין איך לקרוא אותן.

#### מופיע במחברות: 6

</div>

```python
class Queue:
    def __init__(self, lam, mu):
        self.lam = lam
        self.mu = mu

    def utilization(self):
        return self.lam / self.mu

q = Queue(0.8, 1.0)
print(q.lam, q.utilization())
```

## `np.linalg.solve()  inv()  matrix_power()`

### הפקודה `solve` פותרת מערכת משוואות לינאריות `A x = b`.

### במחברת 8 פתרנו כך את משוואות הזרימה של רשת תורים.

### הפקודה `inv` מחשבת מטריצה הפוכה, ו־`matrix_power` מעלה מטריצה בחזקה.

#### מופיע במחברות: 8, 10

```python
from scipy.linalg import inv
from numpy.linalg import matrix_power

# traffic equations: lambda = r + P^T lambda   ->   (I - P^T) lambda = r
P = np.array([[0.0, 0.5],
              [0.2, 0.0]])
r = np.array([1.0, 0.0])
lam = np.linalg.solve(np.eye(2) - P.T, r)
print('lambda =', lam)

A = np.array([[2., 0.], [0., 4.]])
print(inv(A))
print(matrix_power(A, 3))
```

# חלק 6: ספריית simpy — סימולציה מבוססת אירועים

### במחברות 4 ו־6 בנינו סימולציה של תור עם simpy.

### הרעיון: כל לקוח הוא תהליך. התהליך מחכה לזמן (timeout) או למשאב (Resource), ו־simpy מנהלת את השעון.

### המילה `yield` בתוך פונקציה עוצרת את התהליך עד שהאירוע מתרחש, ואז ממשיכה מאותה נקודה.

</div>

<div dir="rtl">

## `simpy.Environment()  env.process()  env.timeout()  simpy.Resource()  env.run()`

### הסביבה `Environment` מחזיקה את שעון הסימולציה, `env.now`.

### הפקודה `process` מפעילה תהליך, ו־`timeout` גורמת לו לחכות זמן נתון.

### המשאב `Resource` הוא השרת. לקוח מבקש אותו עם `request`, ומחכה בתור עד שהוא פנוי.

### המילה `with` דואגת לשחרר את השרת כשהלקוח מסיים. הפקודה `run` מריצה את הסימולציה עד זמן נתון.

#### מופיע במחברות: 4, 6

</div>

```python
import simpy

np.random.seed(0)
waits = []

def customer(env, server, mu):
    arrive = env.now
    with server.request() as req:
        yield req                                          # wait for the server
        waits.append(env.now - arrive)
        yield env.timeout(np.random.exponential(1 / mu))   # service time

def arrivals(env, server, lam, mu):
    while True:
        yield env.timeout(np.random.exponential(1 / lam))  # next arrival
        env.process(customer(env, server, mu))

env = simpy.Environment()
server = simpy.Resource(env, capacity=1)
env.process(arrivals(env, server, lam=0.8, mu=1.0))
env.run(until=20_000)

print('customers served:', len(waits))
print(f'mean wait (simulation): {np.mean(waits):.2f}   M/M/1 theory: {0.8 / (1 * (1 - 0.8)):.2f}')
```
