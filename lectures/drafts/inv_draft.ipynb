## חלק 1: עלות אחזקה
### נניח שהעלות שלנו היא על כל יחידת זמן עבור כל יחידה של מוצר.
### למשל אם תמיד היו 3 מוצרים במלאי אז אני משלם איזשהו סכום קבוע על 3 מוצרים. 
### נהוג לסמן עלות זו ב $C_h$  מהמילה Holding. 
### נסמן את גובה המלאי ברגע אקראי באות $I$ מהמילה Inventory.
### בעצם סה"כ נשלם את $H$ עבור תוחלת גובה המלאי וזה יהיה העלות אחזקה שלנו ההמוצעת שלנו ליחידת זמן

### $$ C_h * E[I] $$

### כאשר $H$ עלות לאחזקת יחידת מלאי אחת ליחידת זמן אחת.
### $E[I]$ תוחלת אחזקת מלאי ליחידת זמן אחת

### אם ידוע ש$H =10$  וממוצע גובה המלאי הוא 5 אז העלות הממוצעת ליחיד זמן אחת היא 5*10 = 50
