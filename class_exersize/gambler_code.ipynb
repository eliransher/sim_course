```python
import numpy as np
from tqdm import tqdm
import matplotlib.pyplot as plt
```

```python
### play until win by one or bankrupt

def play_win_or_lose(p_win, budget):
     
    gamble_val = 1
    while True:
        play_game = np.random.rand()
    
        if play_game <  p_win:
       
            return 1
        else:
            budget -= gamble_val
            
            gamble_val = gamble_val*2
        if budget < gamble_val:
            return 'bankrupt'


    
```

```python
### compute probability of bankrupt


def prob_bankrupt(total_trials, p_win, budget, win_cash_out):
    
    budget_init = budget
    total_wins = 0
    
    for ind in range(total_trials):
        
        budget = budget_init
        while (budget != 'bankrupt') and (budget < 15):
            
        
            cur_game = play_win_or_lose(p_win, budget)
            if cur_game == 1:
                budget += 1
            
        
        if budget != 'bankrupt':
             total_wins += 1
    
    return 1 -  total_wins/total_trials

```

```python
p_win = 0.5
budget = 10
win_cash_out = 15
total_trials = 5000000

prob_bankrupt(total_trials, p_win, budget, win_cash_out)
```

```python
## compute probability of bankrupt for every p

total_trials = 50


prob_bank = []
for p_win in tqdm(np.linspace(0.1, 0.99, 20)):

    budget = 10
    win_cash_out = 15
    prob_bank.append(prob_bankrupt(total_trials, p_win, budget, win_cash_out))
```

```python
plt.figure()
plt.plot(np.linspace(0.1, 0.99, 20), prob_bank)
plt.xlabel('P win')
plt.ylabel('prob bankrupt')
plt.show()
```
